# 1. Load Libraries

```python
import pandas as pd
```

# 2. Load Datasets

```python
df_original = pd.read_csv("../../dataset/dataset.csv")
df_rag = pd.read_csv("../../outputs/RAG/rag_results.csv")

df_original.columns
```

```python
df_rag = df_rag.rename(columns={
    "sentiment": "manual sentiment",
    "text_id": "konten"
})
df_rag["judul"] = "Generated with FLAN-Base"
df_rag.columns
```

# 3. Combine Datasets

```python
df_original["llm_generated"] = False
combine_original_df = df_original[["konten", "manual sentiment", "llm_generated", "judul"]]
```

```python

combine_rag_df = df_rag[["konten", "manual sentiment", "llm_generated", "judul"]]
combine_rag_df
```

```python
df_combined = pd.concat([combine_original_df, combine_rag_df], axis=0, ignore_index=True)
df_combined.to_csv("../../outputs/RAG/dataset_rag_combined.csv", index=False)
```
